Today we willl understand about Functions In python, Function are the special set of progrmas which increase code reusability and makes code cleaner, Indentation is Important Here And Functions Make Specific Tasks Easier When Code Base Gets Huge 

```python
#Let's define a simple function

def greet(name):   # Function Definition
    return f"Hello, {name}!"  

#Let's call the function
print(greet("Akshat"))  # Functuon Call
print(greet("John"))
print(greet("Jane"))

# And This is How We Can Reuse Code With Functions
```

```python
# Let's work With An Example of a Function with Arguments

def avg():
    # Function to calculate the average of two numbers
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    average = (num1 + num2) / 2
    return average

# Let's call the function
result = avg()
print(f"The average is: {result}")
```

Let's test Functions With Arguements

```python
# Functions With Arguements 

def greet(name, age):  # Function Definition with multiple parameters
    return f"Hello, {name}! You are {age} years old."

# Let's call the function
print(greet("Akshat", 25))
print(greet("John", 30))
print(greet("Jane", 28))
```

```python
# now Let's see how we can use default arguments in functions

def greet(name, age=18):  # Function Definition with default parameter
    return f"Hello, {name}! You are {age} years old."

# Let's call the function
print(greet("Akshat"))  # age will take the default value of 18
print(greet("John", 30))  # age will take the provided value of 30
```

```python
# Let's solve Some Practice set Questions 

# Find Greatest Of Three Numbers Using Functions 

def greatest( num1, num2, num3 ):  # Function Definition
    if num1 > num2 and num1 > num3:
        return num1 
    elif num2 > num1 and num2 > num3:
        return num2 
    else:
        return num3

# Let's call the function
greatest_number = greatest(10, 20, 30)
print(f"The Greatest Number Is: {greatest_number}")
```

```python
# Convert Celsius To Fahrenheit Using Functions

def celtofar(celcius):   # Function Definition 
    fahrenheit = (celcius * 9/5) + 32
    return fahrenheit


# let's call The Function 
temperature = celtofar(25)
print(f"25 degrees Celsius is equal to {temperature} degrees Fahrenheit.")
```

```python
# let's take value From User 

def c_to_f(c):  # Function Declaration 
    f = (c * 9/5) + 32
    return f

# To Take Input From User
c = float(input("Enter temperature in Celsius: "))
f = c_to_f(c)  # Function Call
print(f"{c} degrees Celsius is equal to {f} degrees Fahrenheit.")
```
