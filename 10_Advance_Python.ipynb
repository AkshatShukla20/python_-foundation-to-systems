```python
# walrus operator example
# The walrus operator (:=) allows you to assign values to variables as part of an expression. This can make your code more concise and readable.

if (n := len([1, 2, 3, 4, 5])) > 4:
    print(f"list is too long {n} elements expected 4 or fewer")
else :
    print( f"list is acceptable with {n} elements")
```

```python
# One More example of wlarus operator
# The walrus operator can also be used in while loops to read data until a certain condition is met.

while (line := input("Enter a line (or 'quit' to exit): ")) != "quit":
    print(f"You entered: {line}")
    break; 
```

```python
# Type Hinting In Python 

n = 5  # type: int
n2 : int = 10 

def add_numbers(a: int, b: int) -> int:
    return a + b

sum = add_numbers(n, n2)
print(f"The sum of {n} and {n2} is: {sum}")
```

```python
# Advance type hinting in python

from typing import List, Tuple, Dict, Union
# List of integers
numbers: List[int] = [1, 2, 3, 4, 5]
print( f"The list Is {numbers}")
# Tuple of a string and an integer
person: Tuple[str, int] = ("Alice", 30)
print( f" The String is {person}")
# Dictionary with string keys and integer values
scores: Dict[str, int] = {"Alice": 90, "Bob": 85}
print( f"The Scores Are {scores}")
# Union type
identifier: Union[int, str] = "ID123"
```

```python
# Match and Status

def http_status(status):
    match status :
        case 200 :
            print( "OK")
        case 404:
            print( "Error Not Found")
        case 500:
            print( "Internet Connection Issue")
        case _ :
            print("Invalid Case")

print( http_status(500))

         
```

```python
# Merging Dictionaries 

dict1 = { 'a': 3 , 'b' : 4 }
dict2 = { 's' : 10 , 'p' : 11 }

merged = dict1 | dict2
print( merged )
```

```python
# Let's Try with Else Clause 
try :

    a = int(input( "Enter The Number: "))
    print(a) 

except Exception as e :
    print(e)

else :
    print( "you Are In else Clause")
```

```python
# Let's Give it a try with finally keyword 
# Finally has a Main purpose when we deal with functions 

def main() :
    try :
        a = int(input("Enter a number: "))
        print( a )
        return 
    except Exception as e :
        print(e)
        return 

    finally :
        print( "This will print no matter what")
main()
```

```python
# Global Keyword Usage

a = 34

def fun() :
    global a # this will change value of global variable
    a = 3
    print(a)

fun()
print(a)
```

```python
# using enumerates in python 

l = [3,4,5,12,14,7,89]

for index, item in enumerate(1):
    print(f"The item at index {index} in the l is {item}")
```

```python
# List Comprehension In Python 

List = [ 2,3,4,5,6 ]
squared_list = []

for item in List :
    squared_list.append(item*item)

print(squared_list)
# This was Brute method Now Try list comprehension 


squared_list1 = [ i*i for i in List ]
print(squared_list1)
```
