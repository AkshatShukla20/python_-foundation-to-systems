```python
# How To Find Average Of N Numbers In Python 

n = int(input("Enter Number Of Elements:"))
avg = 0
num = 0
i = 1
for i in range(n):
    num = num + i
avg = num / n
print("Average Of N Numbers Is:", avg)
```

```python
# How to find sum of first N positive integers in Python

n = int(input("Enter a range of integers: "))
sum = 0
for i in range(1, n + 1):
    sum += i
print("Sum of first", n, "positive integers is:", sum)
```
