Let's Learn About File Input And Output, And How To Read Text Files In Python 
Open is A Built In function In Python which is used to read files 

```python
f = open("Sample Text File.md", "r")  # Function To Read The File
data = f.read()   # Necessary Syntax 
print(data)
f.close()  # Closure of the File
```

```python
# to Write a file we use "w" instead of "r"

st = "I Know I'm GOAT"

f = open("Sample Text File.md", "w")  # Function To Write The File
f.write(st)
f.close()  # Closure of the File
```

```python
# Now let's test more file operations like append, readlines, etc.

f = open( "Sample Text File.md" )
data = f.readlines()  # Readlines will read the file line by line and return a list of lines
print(data)  # returns the list of lines in the file
print( type(data))  # It will return the type of data which is list
f.close()  # Closure of the File
```

```python
f = open("Sample Text File.md", "a")  # Function To Append The File
f.write("\nI am the GOAT")  # Appending a new line to the file 
f.close()  # Closure of the File
```

```python
# Read In Binary Mode And read In text mode 

f = open("Sample Text File.md", "rb")  # Function To Read The File In Binary Mode
data = f.read()   # Necessary Syntax
print(data)
f.close()  # Closure of the File  

f = open("Sample Text File.md", "rt")  # Function To Read The File In Text Mode
data = f.read()   # Necessary Syntax
print(data)
f.close()  # Closure of the File  
```

```python
# let's use with statement to open a file, it will automatically close the file after the block of code is executed

with open("Sample Text File.md", "r") as f:  # Function To Read The File
  print(f.read()) #Indentation Is Used Here Too
```
