```python
# Let's Start By Writing A Program For Factorial Calculation

def factorial(n):
    if n < 0:
        return "Invalid Number"
    elif n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

# Taking Input From User
n = int(input("Enter The Number To Calculate Factorial: "))
print(f"The Factorial Of {n} Is: {factorial(n)}")
```

```python
# Now Let's Calculate Sum Of First n Numbers Using Recursion

def sum(n):
    if n<= 0:
        return 0
    else:
        return n + sum(n - 1)

# Taking Input From User
n = int(input("Enter The Number To Calculate Sum: "))
print(f"The Sum Of First {n} Numbers Is: {sum(n)}")
```

```python
# let's Draw a pattern using recursion

def pattern(n):
    if n == 0:
        return
    print("*" * n)
    pattern(n - 1)

# Taking Input From User
n = int(input("Enter The value: "))
print(f"Pattern For {n} Is:")
pattern(n)
```
