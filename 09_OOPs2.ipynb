Now We Will Discuss More About The OOPs..
So There Are Four Main Pillars Of OOPs Which Are 
1. Encapsulation- Wrapping up of data into one single self contained unit 
2. Abstraction - Practice of hiding complex details to user and expose only essential features 
3. Inheritance - A mechanism that allows sub-class to inherit properties, behaviour from super-class
4. Polymorphism - An Ability of an different object to act like object of same class through common interface 

```python
# Concept Of Inheritance in Python

class Employee:   # this Is Parent Class
    def __init__(self, name, salary, company): 
        self.name = name 
        self.salary = salary
        self.company = company
        print (f"The name of the employee is {self.name} and his salary is {self.salary} and he works in {self.company}")


class SoftwareEngineer(Employee):   # This is Sub_Class 
    def __init__(self, name , salary , company, programming_language):
        super().__init__(name, salary, company)  # Calling the constructor of the parent class
        self.programming_language = programming_language
        print(f"The programming language of the software engineer is {self.programming_language}")


Akshat = SoftwareEngineer("Akshat", 50000, "Google", "Python")  # Creating an object of the subclass
Sahil = SoftwareEngineer("Sahil", 60000, "Microsoft", "Java")  # Creating another object of the subclass
```

```python
#  Now we will work on concept of multiple inheritance in python

class Person:   # this Is Parent Class
    def __init__(self, name, age): 
        self.name = name 
        self.age = age
        print (f"The name of the person is {self.name} and his age is {self.age}")

class Profession:
    def __init__(self, profession): 
        self.profession = profession
        print (f"The profession of the person is {self.profession}")

class SoftwareEngineer(Person, Profession):
    def __init__(self, name, age, profession):
        Person.__init__(self, name, age)
        Profession.__init__(self, profession)

Akshat = SoftwareEngineer("Akshat", 25, "Software Engineer")  # Creating an object of the subclass
```

```python
# Class Method Overriding in Python 

class Student:  # this is parent class 
    def __init__(self, name, age, section):
        self.name = name
        self.age = age 
        self.section = section 
        print(f"The name of student is {self.name} and He is {self.age} years old and He is in {self.section} section")

class CollegeStudent(Student):  # this is subclass
    def __init__(self, name, age, section, college_name):
        super().__init__(name, age, section)  # Calling the constructor of the parent class
        self.college_name = college_name
        print(f"The name of the college is {self.college_name}")

class Details( CollegeStudent , Student): 
    @classmethod  # This is decorator which is used to define a class method
    def student_details(cls, name, age, section, college_name):
        return cls(name, age, section, college_name)  # This will call the constructor of the subclass


Akshat = Details.student_details("Akshat", 20, "A", "IIT Kanpur")  # Creating an object of the subclass using class method
```

```python
# Property Decorators in Python, Setters and Getters in Python

class Names:
    @property  # This is decorator which is used to define a property method
    def name(self):
        return self._name  # This will return the value of the private variable _name   

    @name.setter  # This is decorator which is used to define a setter method
    def name(self, name): 
        self._name = name.split()  # This will split the name into first name and last name and store it in the private variable _name
        self.first_name = self._name[0]  # This will store the first name in the variable first_name
        self.last_name = self._name[1]  # This will store the last name in the variable last_name

Akshat = Names() # Creating an object of the class
Akshat.name = "Akshat Shukla" # This will call the setter method and set the value of the private variable _name
print(Akshat.name) # This will call the property method and print the value of the private variable _name
print(Akshat.first_name) # This will print the first name
print(Akshat.last_name) # This will print the last name


#These Overriding methods Are Part Of Abstraction And Encapsulation In Python
```

```python
# Last Part Operator Overloading In Python 

class Operations:
    def __init__(self, a, b):
        self.a = a
        self.b = b  

    def __add__(self, other):  # This is operator overloading method for addition
        return self.a + other.a, self.b + other.b  # This will return the sum of the two objects


Sum = Operations (10, 20)  # Creating an object of the class
Sum1 = Operations (30, 40)  # Creating another object of the class
print(Sum + Sum1)  # This will call the __add__ method and print the sum of the two objects
```
