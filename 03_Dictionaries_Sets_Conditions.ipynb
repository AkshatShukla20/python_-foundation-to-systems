Now We Will Understand Dictionaries And Sets Along With Conditional Statements/ Expressions In Python With Few Examples 
Dictionary Is Like Pair Of Key : Value List
Need :- Lookup Is So fast It Gives O(1), Even If List Has 1000 Students 
It Is Mutable, un-ordered, Can't Duplicate Keys, Indexed thats Why It Is fast For Lookup 

```python
# Demonstration Of Python Code for Dictionaries 
marks = {
    "Akshat" : 99,
    "Sahil" : 95,
    "Rohit" : 92,
}
print ( type(marks) ) #This will print the type of marks which is dictionary
print ( marks ) #This will print the entire dictionary
d = {} #This will create an empty dictionary
```

```python
# Let's Test Some Dictionary Methods
print ( marks.keys() ) #This will print all the keys of the dictionary
print ( marks.values() ) #This will print all the values of the dictionary
print ( marks.items() ) #This will print all the key-value pairs of the dictionary
marks.update({"Rohit" : 100}) #This will update the value of Rohit to 100
print ( marks ) #This will print the updated dictionary
print ( marks.get("Aastha"))  #This will print None because the key "Aastha" does not exist in the dictionary
```

Key Difference Between marks.get and marks[] is that marks.get will give "none" while marks[] will throw error.
Use Documentation For More Methods.

Now Let's Jump To Sets, Just Like Maths Here too Set is A Collection Of Well Defined Objects 
Every Value Must Be Unique, It Is Un-Ordered, Un-Indexed, Set Is Muttable But catch Is We Can't manipulate existing Values 

```python
# Demonstration Of Python Code for Sets
s = { 1, 2, 3, 4, 5 }
s2 = set()  # This will create an empty set
print ( type(s) ) #This will print the type of s which is set
print ( s2 ) #This will print the empty set
s.add(6) #This will add 6 to the set
print ( s ) #This will print the entire set
s.difference_update({ 1, 2, 3 }) #This will remove 1, 2 and 3 from the set
print ( s ) #This will print the updated set
s.clear() #This will remove all the elements from the set
print ( s ) #This will print the empty set
print ( len(s) ) #This will print the length of the set which is 0 
```
