```python
# Map, filter, Reduce 
# map exmaple
from functools import reduce # to Import reduce 

l =[ 1, 2, 3, 4, 5]
square = lambda x : x*x
sqList = map(square, l)

print(list(sqList))

# filter example 

def even (n):
    if ( n % 2 == 0) :
        return True
    return False
onlyEven = filter(even, l)
print(list(onlyEven))

# Reduce examples 
# reduce add the list - first it will add 1,2 then result added to 3 then to 4 lastly 5

def sum(a, b):
    return a + b  
print( reduce( sum, l ))
```
