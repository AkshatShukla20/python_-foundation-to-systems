From Here We Will Start The Tricky Portion Of Every Programming language, Strings 
1. Strings Are Immutable {once declared, then can't be changed}
2. Indexing Starts From 0, And we Can Treat Letters Array {From 0 to Length-1}

```python
# Now Let's Declare String Methods 

name = "akshat"
name_ = name.capitalize()  # Capitalizes the first letter of the string
print ( name_)
 # Ways to declare a string
 # "double quotes"
 # 'single quotes' 
 # '''triple quotes'''

 # To Cut string 
name_2 = name[0:5]  # Extracts characters from index 0 to 4
print(name_2)

# Negative Indexing
name_3 = name[-1]  # Accesses the last character of the string
print(name_3)

# Usually We Never Use Negative Indexing Method 
# Now Let's Use Skipping Slice Method 

name_4 = "sahil_nigger"
name_5 = name_4[0:10:2]  # Extracts characters from index 0 to 9 with a step of 2
print(name_5) # Here 3rd Value Decides the Skip eg. 2 Here 
```

Now Let's Try With String Functions, Although There Are So Many But Let's start With Basic Ones 

```python
cand = "Akshat"
print(len(cand)) #Length Of String
print(cand.endswith("t")) #returns boolean value
print(cand.startswith("A")) #returns boolean value
print(cand.lower()) # Convert To Lowercase
print(cand.upper())  # Convert To Uppercase
print(cand.count("a")) #Counts The Exact variable Repeated When And How many Times 
```

Now We Will Start With Lists And Tuples, List Is A Part In STL Of C++
Just Like In Dart And C++ Lists Can Store Values Of Any Data Type, we Can Counter String Immutibility By Using Lists 

```python
# Example of breaking immutibility of string 
# Declaration Of List 
my_list = ["Rohan", "Vishal", 69, 78.90, True]
print(my_list[1])
my_list[1] = "Aman"
print(my_list[1])  # Demonstration Of List Mutable
```

```python
#Test Of List Methods 

l1 = [ 1,72, 69, 56, 2, 33]
l1.sort() # Sort The List
print(l1)
l2 = [ 4, 8, 12, 16, 20 ]
l2.reverse() # reverse The List
print(l2)
l3 = [ 23, 34, 45, 56, 67 ]
l3.insert(2, 40) # insert 40 at index 2
print(l3)
l4 = [ 10, 20, 30, 40, 50 ]
print(l4.pop(2))  # POP and Print The Value At Index 2
print(l4)  # Print The List After Popping
l4.remove(40)  # Remove The Value 40 From The List
print(l4)  # Print The List After Removing 40
```

Now Using Tuples :- They Are Like Lists But Not Immutible, Means Data Can Not Be Changed 


```python
# Declaration Of Tuple
my_tuple = (1, 2, 3, 4, 5)
# Methods Of Tuple, There Are Only 2 Methods Of Tuple
print(my_tuple.count(2))  # Counts The Occurrences Of 2 In The Tuple
print(my_tuple.index(3))  # Finds The Index Of 3 In The Tuple

# Operations On Tuples, There Are So Many Operations....Use Documentation For More Operations
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
# Concatenation
tuple3 = tuple1 + tuple2
print(tuple3)  # Output: (1, 2, 3, 4, 5, 6)
# Repetition
tuple4 = (7, 8, 9)
tuple5 = tuple4 * 3
print(tuple5)  # Output: (7, 8, 9, 7, 8, 9, 7, 8, 9)
```
