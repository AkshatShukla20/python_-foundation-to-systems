Type() And Type Casting, already Know It Is Explicit Conversion of Data Type From One To Another 
Type() Is Used To Display Datatype Of Variable. Let's Put It To Test 

```python
a = 10 
t = type( a ) #class<int>
print( t ) # This will print the data type of variable a
b = 3.14
t = type( b ) #class<float>
print( t ) # This will print the data type of variable b
c = "69.3"
t = type( c ) #class<string>
print( t ) # This will print the data type of variable c
```

```python
# Testing Type Casting in Python
a = 31
b = float( a ) # This will convert the integer a to a float
print( b ) # This will print the float value of variable a
t = type (b) #This will get the data type of variable b
print( t ) # This will print the data type of variable b

c = 21
d = str( c ) # This will convert the integer c to a string
print( d ) # This will print the string value of variable c
t_ = type (d)
print( t_ ) # This will print the data type of variable d
```
