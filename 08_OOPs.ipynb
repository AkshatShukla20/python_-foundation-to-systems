Today We Will Start by Learning Object oriented Programming, It is a paradigm of programming which organize software design around objects instead of Functions, logic. It Promotes Code Reusibility  {DRY Practice}.
There Are Two main Components Of OOPs 
1. Class - Blueprint to create an object, it defines object behaviour, attribute, structure  
2. Object - Special Instance Of Class, which has some actual data in it 

```python
# let's create our first class in python 
class Student:
    #constructor
    def __init__(self):   # self is a reference to the current instance of the class
        self.name = "akshat"
        self.branch = "CSE"    # Class attributes are variables that are shared among all instances of a class. 
                              # They are defined within the class but outside any methods. In this case, name and branch are instance attributes, which means they are unique to each instance of the Student class   
        self.roll_no = 300

Physics = Student()
print(Physics.name)
print(Physics.branch)
print(Physics.roll_no)
```

```python
# Get Info Logic 
class Workers:
    def __init__(self, name, age, salary):  # init is used to initialize attribute of class

        name = input("Enter your name: ")
        salary = input("Enter your salary: ")
        age = input("Enter your age: ")
        self.name = name
        self.age = age
        self.salary = salary

    def get_info(self):
        print(f"Name: {self.name}, Age: {self.age}, Salary: {self.salary}")
 
```

```python
# instance vs class attributes

class workforce:
    language = "Embedded C"  # class attribute
    salary = 1200000  # Rich Heheh

Akshat = workforce()
Akshat.language = "Python"  # instance attribute
print(Akshat.language)  # Output: Python

# Logic :- Instance Attributes Is Preferred Over Class Attributes If Both Have Same Name.
```

```python
# More Get Info Logic

class Employee:
    language = "Python"  # class attribute
    salary = 1200000  # class attribute 

    def get_info(self):
        print(f"Language: {self.language}, Salary: {self.salary}")

    def greet(self):
        print(f"Hello, I am an employee who knows {self.language} and earns {self.salary}.")


Akshat = Employee() 
Akshat.greet()  # Output: Hello, I am an employee who knows Python and earns 1200000.
Akshat.get_info()  # Output: Language: Python, Salary: 1200000
```

```python
# static method 

class MathOperations:
    @staticmethod   # Static methods do not require objects 
    def add(a, b):
        return a + b

    @staticmethod
    def multiply(a, b):
        return a * b

# Usage
result_add = MathOperations.add(5, 3)
result_multiply = MathOperations.multiply(5, 3)
print(result_add)  # Output: 8
print(result_multiply)  # Output: 15
```

```python
# Constructor - __init__ method

class Person:
    def __init__(self, name, age):  # Constructor is a special method that is automatically called when an object of the class is created. It is used to initialize the attributes of the class. The __init__ method takes self as the first parameter, which refers to the instance of the class being created. Additional parameters can be added to the __init__ method to allow for customization of the object's attributes during instantiation.
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}")  

Akshat = Person("Akshat", 20)
Akshat.display_info()  # Output: Name: Akshat, Age:

# dunder methods are automatically called when an object is created in class. 
```
