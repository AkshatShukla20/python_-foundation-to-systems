Today we Will Learn About Conditionals In Python Which Are Conditional Statements Like In C++ But With A Change 
Here If, Else, Elif Is Used instead of Else if


```python
# Conditional Statements Structure 

name = input("Enter Your Name:")
age = int( input( "Enter Your Age : "))

if( age >= 18) :
    print(name + " is eligible to vote")  # This Space After : In If IS Called Indentation
```

```python
# Now Lets Do All 3 Together 

age = int( input( "Enter Age Fam : "))
if( age > 18) :
    print( "Eligible To Vote")
elif( age < 0):
        print( "Invalid Age")
else :
     print( "Not Eligible To Vote")


print( " End Of Program")

```

```python
# can Use Elif Clause, Just Like Nested If Else In C++

marks = int( input( "Enter Your Marks"))
if( marks >= 80):
    print( "A+")
elif( marks < 80 and marks >= 60):  # Usage Of Logical Operators
    print( "A")

elif( marks < 60):
    print( "B")

elif( marks < 40 and marks > 20 ):
    print( "C")

else :
    print( "Fail")
        
```

Flow Control - Conditional statements We say Which Has looping Statements, But With Good Change In Syntax
1. For Loop
2. While Loop 

```python
#Let's Test For Loop First, The Syntax Is Quite Different 

for i in range ( 1, 6) : 
    print(i)
```

```python
# for While Loop This Is The Code and Conditions 

i = 1 # Value initialization 
while( i <= 10) :   # Condition 
      print (i)     # Execution 
      i += 1
```

```python
# Can be used for skipping steps as well, A list tuple and string can be iterate using For loop 

i = 0
for i in range ( 1, 10, 2):
    print(i)
    i += 1
```

```python
# content of list can be print using while loops 

l = [1, "Harsh", False, "Femcel", 23.2]
i = 0
while(i < len(l)):
    print(l[i])
    i += 1
```

```python
# usage of for loop with else { happens with for loop is exhausted }
l = [ 1, 2, 69, 4, 5]
i = 0
for i in l:
    print( i )
    i += 1

else :
    print( "Done")
```

```python
# Now Let's Talk About Usage Of Break, Continue, Pass 
# Break Statement

for i in range(10):
    
    if( i == 5) : # Usage of relational operator 
        break # Completely exit Loop After condition Satisfies
    print ( i )
    i += 1 

# Continue Statement

for p in range(10) :
    if ( p == 7):
        continue # Skips the particular iteration 
    
    print( p )


# Pass Statement- Null Statement 
for s in range(2000) :
    pass # this will let us focus on while loop without taking care of for loop here

q = 0
while ( q < 10):
    print( q )
    q += 1
```
